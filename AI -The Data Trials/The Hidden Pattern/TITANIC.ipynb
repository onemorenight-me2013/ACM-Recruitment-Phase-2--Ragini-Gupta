```python
import pandas as pd
df=pd.read_csv("train.csv")
#  grouping by pclass (1,2,3) and aggreating it to making a new column for mean fare , age and survied rates to save the mean of these values of each grop
g=df.groupby("Pclass").agg( mean_fare=("Fare","mean"),mean_age=("Age","mean"),survied_rate=("Survived","mean"), count=("PassengerId","count")).round(2)
print(g)
```
