```python
import pandas as pd
df=pd.read_csv("train.csv") # to read the csv file
print(df.isnull().sum()) # seeing if any null values are there or not
df_drop=df.dropna()# droping the missing values
print(" original shape:",df.shape)
print("after dropna:",df_drop.shape)
```

```python
# filling the missing values ( if any)
df["Age"]=df["Age"].fillna(df["Age"].mean()) # by writing mean() we replace it with actual number instead of method object(in mean)
df["Cabin"]=df["Cabin"].fillna("Unknown")
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])# we wrote mode()[0] so that we can get a single value otherwise it gives 0 s as output
print(df.isnull().sum()) # checking if any null values are there or not
```

```python
print(df.duplicated().sum()) # checking if any duplicates values are there or not
df_duplicate=df.drop_duplicates() # dropping the duplicate values if any
print(df.shape) # getting the shape before dropping the duplicate values
print(df_duplicate.shape)# getting shape after dropping the duplicate values
```

```python
names=df.loc[df["Age"] > 60 , "Name"] # printing the names of persons whose age is greater than 60
print(names)
```

```python
print(df.iloc[0:5 , 0:4]) # getting first five columns ( from 0 to 4) and first 4 rows (  from 0 to 3)
```

```python
df.to_csv("train_Cleaned.csv", index=False)
print("Saved cleaned dataset successfully!")# saved the cleaned data successfully for futher use
```
