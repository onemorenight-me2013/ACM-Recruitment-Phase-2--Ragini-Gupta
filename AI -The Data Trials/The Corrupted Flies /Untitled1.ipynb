```python
import pandas as pd
df=pd.read_csv("student_messy_data.csv")
#replacing the irrgeular values of gender
df["Gender"]=df["Gender"].replace({"f":"Female" , "F":"Female"})
df["Gender"]=df["Gender"].replace({"m":"Male" , "M":"Male"})
```

```python
print("number of duplicate values:",df.duplicated().sum())# number of duplicate values
df_duplicate=df.drop_duplicates()# dropping the duplicate values if any
print("shape before removing duplicate values:",df.shape)
print("shape after removing duplicate values:",df_duplicate.shape)
```

```python
# replacing the messy marks values with null values but type would be an integer
df["Math"] = pd.to_numeric(df["Math"], errors="coerce").astype("Int64")
df["Science"] = pd.to_numeric(df["Science"], errors="coerce").astype("Int64")
df["English"] = pd.to_numeric(df["English"], errors="coerce").astype("Int64")
```

```python
print(df.isnull().sum())# null values
# if then fill it with median/mean values since they are null values but can drop or delete them
df["Math"]=df["Math"].fillna(df["Math"].median())
df["Science"]=df["Science"].fillna(df["Science"].median())
df["English"]=df["English"].fillna(df["English"].median())
print("after filling missing values:", df.isnull().sum())
```

```python
df.to_csv("students_messy_data_cleaned.csv",index=False)
print('saved clean dataset successfully')
```
