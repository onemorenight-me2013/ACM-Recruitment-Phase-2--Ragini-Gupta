```python
import pandas as pd
df=pd.read_csv("train.csv")
data={"Embarked":["C","Q","S"],"PortName":["Cherboung","Queenstown","Southampton"]}#creating a list tp store these port's name
data_frame=pd.DataFrame(data)#creating a tabular form forvthese port in the dataset
merged=df.merge(data_frame,on="Embarked",how="left")#merging this table with the dataset table
print(merged[["PassengerId","Name","Embarked","PortName"]].head(8))# displaying only the details which is asked to print upto 8th column and row
merged.to_csv("train_with_ports.csv",index=False)# saving the information
print("saved dataset succesfully")
```
